```python
import nltk
from nltk.corpus import gutenberg
text = gutenberg.raw('austen-emma.txt')

# helper functions from chapter 2
def bigram_count(text:str):
    corpus = text.split()
    bi_counter = {}
    n =2
    for i in range(len(corpus) - n +1 ):
        window = corpus[i:i+n]
        history = tuple(window[:1])
        next_word = window[1]
        if history not in bi_counter:
            bi_counter[history] = {next_word:1}
        else:
            if next_word not in bi_counter[history]:
                bi_counter[history][next_word] = 1 
            else:
                bi_counter[history][next_word] += 1
    return bi_counter

def smoothed_prob(history, word, counter, k ,vocab_size):
    next_word_counts = counter.get(history, {})
    count_hw = next_word_counts.get(word,0)
    count_h = sum(next_word_counts.values())
    P_k = (count_hw + k) / (count_h + k*vocab_size)
    return P_k

```
